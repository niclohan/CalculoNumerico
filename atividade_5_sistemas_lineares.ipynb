## Parte 1 — Conceitos e resolução manual

### Exercício 1 — Sistemas na forma matricial $Ax = b$

A ideia é separar o sistema em três partes:
- $A$: matriz dos **coeficientes** (cada linha é uma equação, cada coluna é uma incógnita);
- $x$: vetor das **incógnitas**;
- $b$: vetor dos **termos independentes** (o que está depois do `=`).

**a)** $\begin{cases} 2x + y = 7 \\ x - 3y = -5 \end{cases}$

$$
\underbrace{\begin{bmatrix} 2 & 1 \\ 1 & -3 \end{bmatrix}}_{A}
\underbrace{\begin{bmatrix} x \\ y \end{bmatrix}}_{x}
=
\underbrace{\begin{bmatrix} 7 \\ -5 \end{bmatrix}}_{b}
$$

**b)** $\begin{cases} x + 2y - z = 2 \\ 3x - y + 2z = 9 \\ 2x + y + z = 7 \end{cases}$

$$
\underbrace{\begin{bmatrix} 1 & 2 & -1 \\ 3 & -1 & 2 \\ 2 & 1 & 1 \end{bmatrix}}_{A}
\underbrace{\begin{bmatrix} x \\ y \\ z \end{bmatrix}}_{x}
=
\underbrace{\begin{bmatrix} 2 \\ 9 \\ 7 \end{bmatrix}}_{b}
$$

### Exercício 2 — Eliminação de Gauss (à mão)

Sistema:
$$\begin{cases} 2x + y - z = 8 \\ -3x - y + 2z = -11 \\ -2x + y + 2z = -3 \end{cases}$$

**Passo 0 — matriz aumentada** $[A \mid b]$:

$$\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ -3 & -1 & 2 & -11 \\ -2 & 1 & 2 & -3 \end{array}\right]$$

**Passo 1 — zerar a coluna 1 abaixo do pivô $2$.**
Multiplicadores: $m_{21} = \frac{-3}{2} = -1{,}5$ e $m_{31} = \frac{-2}{2} = -1$.

- $L_2 \leftarrow L_2 + 1{,}5\,L_1$
- $L_3 \leftarrow L_3 + 1\,L_1$

$$\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ 0 & 0{,}5 & 0{,}5 & 1 \\ 0 & 2 & 1 & 5 \end{array}\right]$$

**Passo 2 — zerar a coluna 2 abaixo do pivô $0{,}5$.**
Multiplicador: $m_{32} = \frac{2}{0{,}5} = 4$.

- $L_3 \leftarrow L_3 - 4\,L_2$

$$\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ 0 & 0{,}5 & 0{,}5 & 1 \\ 0 & 0 & -1 & 1 \end{array}\right]$$

Essa é a matriz triangular superior.

**Passo 3 — substituição regressiva** (de baixo para cima):

- $L_3$: $-z = 1 \Rightarrow z = -1$
- $L_2$: $0{,}5y + 0{,}5z = 1 \Rightarrow 0{,}5y = 1 + 0{,}5 = 1{,}5 \Rightarrow y = 3$
- $L_1$: $2x + y - z = 8 \Rightarrow 2x = 8 - 3 - 1 = 4 \Rightarrow x = 2$

$$\boxed{x = 2,\quad y = 3,\quad z = -1}$$

*Conferência na 1ª equação:* $2(2) + 3 - (-1) = 8$ ✓

## Parte 2 — Propriedades e interpretação

### Exercício 3 — Número de soluções

Reduzimos a matriz aumentada e olhamos para a **última linha**:

| Última linha fica assim... | Significa |
|---|---|
| $[\,0 \ \ 0 \mid 0\,]$ | uma equação "sobra" (é redundante) → **infinitas soluções** |
| $[\,0 \ \ 0 \mid c\,]$, com $c \neq 0$ | equação impossível ($0 = c$) → **nenhuma solução** |
| pivô não nulo em todas as colunas | **solução única** |

**a)** $\begin{cases} x + y = 3 \\ 2x + 2y = 6 \end{cases}$

$$\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 6 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 0 \end{array}\right]$$

A segunda equação virou $0 = 0$: era só a primeira multiplicada por 2. Sobra uma incógnita livre
($x = 3 - y$ para qualquer $y$). → **Infinitas soluções.**

**b)** $\begin{cases} x + y = 3 \\ 2x + 2y = 7 \end{cases}$

$$\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 7 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 1 \end{array}\right]$$

A segunda equação virou $0 = 1$, que é impossível. → **Nenhuma solução.**

**c)** $\begin{cases} x + y = 3 \\ 2x - y = 0 \end{cases}$

$$\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & -1 & 0 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & -3 & -6 \end{array}\right]$$

Dois pivôs ($1$ e $-3$), nenhum nulo. $-3y = -6 \Rightarrow y = 2$ e $x = 3 - 2 = 1$. → **Solução única:** $(x, y) = (1, 2)$.

**Conferência em código.** Um jeito rápido de checar: comparar o *posto* (rank) de $A$ com o de $[A \mid b]$.

- posto$(A)$ = posto$([A|b])$ = número de incógnitas → solução única
- posto$(A)$ = posto$([A|b])$ < número de incógnitas → infinitas soluções
- posto$(A)$ < posto$([A|b])$ → nenhuma solução

```python
def classifica(A, b):
    A = np.array(A, dtype=float)
    Ab = np.column_stack([A, b])          # matriz aumentada [A | b]
    rA = np.linalg.matrix_rank(A)
    rAb = np.linalg.matrix_rank(Ab)
    n = A.shape[1]                        # número de incógnitas

    if rA < rAb:
        return "nenhuma solução"
    if rA == n:
        return "solução única"
    return "infinitas soluções"

print("a)", classifica([[1, 1], [2, 2]], [3, 6]))
print("b)", classifica([[1, 1], [2, 2]], [3, 7]))
print("c)", classifica([[1, 1], [2, -1]], [3, 0]))
```

## Parte 3 — Implementação numérica em Python

### Exercício 4 — Eliminação de Gauss sem pivoteamento

O algoritmo tem 3 etapas, e o código abaixo segue exatamente essa ordem:

1. **Matriz aumentada:** juntar $A$ e $b$ numa matriz só, $[A \mid b]$.
2. **Eliminação:** para cada coluna $k$, usar a linha $k$ (o pivô) para zerar tudo que está abaixo dela.
   Para a linha $i$, o multiplicador é $m = \dfrac{a_{ik}}{a_{kk}}$ e fazemos $L_i \leftarrow L_i - m\,L_k$.
3. **Substituição regressiva:** com a matriz triangular, resolver da última linha para a primeira.

"Sem pivoteamento" significa que **não trocamos linhas**. Por isso, se algum pivô for zero, o algoritmo
não consegue continuar — o código avisa com um erro em vez de dividir por zero.

```python
def gauss(A, b, mostrar_passos=False):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # 1) matriz aumentada [A | b]
    M = np.column_stack([A, b])
    if mostrar_passos:
        print("Matriz aumentada:\n", M, "\n")

    # 2) eliminação: zera os elementos abaixo de cada pivô
    for k in range(n - 1):
        if abs(M[k, k]) < 1e-12:
            raise ValueError(f"Pivô nulo na posição ({k},{k}): sem pivoteamento não dá para continuar.")
        for i in range(k + 1, n):
            m = M[i, k] / M[k, k]             # multiplicador
            M[i, :] = M[i, :] - m * M[k, :]   # L_i <- L_i - m * L_k
        if mostrar_passos:
            print(f"Depois de zerar a coluna {k + 1}:\n", M, "\n")

    # o último pivô também precisa ser não nulo
    if abs(M[n - 1, n - 1]) < 1e-12:
        raise ValueError("Último pivô nulo: o sistema não tem solução única.")

    # 3) substituição regressiva: da última linha para a primeira
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = M[i, i + 1:n] @ x[i + 1:n]     # termos das incógnitas já conhecidas
        x[i] = (M[i, n] - soma) / M[i, i]

    return x
```

**Teste com o sistema do Exercício 2** (mostrando a matriz aumentada em cada etapa, como feito à mão):

```python
A = [[2, 1, -1],
     [-3, -1, 2],
     [-2, 1, 2]]
b = [8, -11, -3]

x_gauss = gauss(A, b, mostrar_passos=True)
print("Solução (nossa função):  ", x_gauss)
```

**Comparação com `numpy.linalg.solve`:**

```python
x_numpy = np.linalg.solve(A, b)

print("Solução (numpy):         ", x_numpy)
print("Diferença máxima:        ", np.max(np.abs(x_gauss - x_numpy)))
print("São iguais?              ", np.allclose(x_gauss, x_numpy))
```

**Interpretação.** As duas soluções coincidem com a que obtivemos à mão no Exercício 2
($x=2,\ y=3,\ z=-1$); a diferença é da ordem do erro de arredondamento do computador (praticamente zero).
O `numpy.linalg.solve` usa uma variante da eliminação de Gauss **com** pivoteamento, que é mais estável
numericamente, mas para este sistema (pivôs $2$, $0{,}5$ e $-1$, todos não nulos) o resultado é o mesmo.

### Exercício 5 — Modelo de balanço térmico

$$
\begin{bmatrix} 4 & -1 & 0 \\ -1 & 4 & -1 \\ 0 & -1 & 3 \end{bmatrix}
\begin{bmatrix} T_1 \\ T_2 \\ T_3 \end{bmatrix} =
\begin{bmatrix} 15 \\ 10 \\ 10 \end{bmatrix}
$$

**a) Resolução manual**

Matriz aumentada:
$$\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ -1 & 4 & -1 & 10 \\ 0 & -1 & 3 & 10 \end{array}\right]$$

**Passo 1** — pivô $4$. Multiplicador $m_{21} = \frac{-1}{4}$, e $m_{31} = 0$ (já é zero).
$L_2 \leftarrow L_2 + \frac{1}{4}L_1$:

$$\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ 0 & 3{,}75 & -1 & 13{,}75 \\ 0 & -1 & 3 & 10 \end{array}\right]$$

**Passo 2** — pivô $3{,}75$. Multiplicador $m_{32} = \frac{-1}{3{,}75} = -\frac{4}{15}$.
$L_3 \leftarrow L_3 + \frac{4}{15}L_2$:

- elemento $(3,3)$: $3 + \frac{4}{15}(-1) = \frac{41}{15}$
- termo independente: $10 + \frac{4}{15}(13{,}75) = 10 + \frac{11}{3} = \frac{41}{3}$

$$\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ 0 & 3{,}75 & -1 & 13{,}75 \\ 0 & 0 & \frac{41}{15} & \frac{41}{3} \end{array}\right]$$

**Substituição regressiva:**

- $\frac{41}{15}T_3 = \frac{41}{3} \Rightarrow T_3 = 5$
- $3{,}75\,T_2 - T_3 = 13{,}75 \Rightarrow 3{,}75\,T_2 = 18{,}75 \Rightarrow T_2 = 5$
- $4T_1 - T_2 = 15 \Rightarrow 4T_1 = 20 \Rightarrow T_1 = 5$

$$\boxed{T_1 = T_2 = T_3 = 5}$$

*Conferência na 2ª equação:* $-5 + 4(5) - 5 = 10$ ✓

**b) Implementação em Python** (reaproveitando a função `gauss` do Exercício 4):

```python
A_termica = [[4, -1, 0],
             [-1, 4, -1],
             [0, -1, 3]]
b_termica = [15, 10, 10]

T = gauss(A_termica, b_termica)
print("T1, T2, T3 =", T)
print("numpy       =", np.linalg.solve(A_termica, b_termica))
```

**Interpretação.** O código dá $T_1 = T_2 = T_3 = 5$, igual ao cálculo manual. Esse tipo de sistema
(diagonal dominante: o termo da diagonal é maior que a soma dos outros da linha) é bem comportado, então a
eliminação sem pivoteamento funciona sem problemas. Fisicamente, cada equação diz que a temperatura de um
ponto é influenciada pelas dos vizinhos mais uma fonte externa; a solução mostra que a rede chega ao equilíbrio
com os três pontos na mesma temperatura.
