### Métodos da Bisseção, de Newton e da Secante

Nome: [seu nome aqui]

Neste notebook implemento os três métodos para encontrar raízes de $f(x) = 0$ e aplico
em quatro funções diferentes, comparando os resultados.

**Critério de parada:** paro a iteração quando a diferença entre duas aproximações
seguidas fica menor que a tolerância ($|x_{k+1}-x_k| < tol$) ou quando o valor da
função já está bem perto de zero ($|f(x_{k+1})| < tol$). Uso tol = $10^{-6}$ e no
máximo 100 iterações.

## Bibliotecas

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

TOL = 1e-6
MAX_ITER = 100
```

## Parte 1 — Implementação dos métodos

Cada função devolve a raiz encontrada, quantas iterações foram usadas, a lista com
todas as aproximações (para poder plotar depois) e se o método convergiu ou não.

```python
def bissecao(f, a, b, tol=TOL, max_iter=MAX_ITER):
    if f(a) * f(b) > 0:
        print("Erro: f(a) e f(b) tem o mesmo sinal, não dá pra usar bisseção nesse intervalo")
        return None, 0, [], False

    aproximacoes = []
    for i in range(max_iter):
        m = (a + b) / 2
        aproximacoes.append(m)

        if abs(f(m)) < tol or (b - a) / 2 < tol:
            return m, i + 1, aproximacoes, True

        if f(a) * f(m) < 0:
            b = m
        else:
            a = m

    return m, max_iter, aproximacoes, False
```

```python
def newton(f, df, x0, tol=TOL, max_iter=MAX_ITER):
    x = x0
    aproximacoes = [x0]

    for i in range(max_iter):
        if abs(df(x)) < 1e-12:
            print("Erro: derivada muito perto de zero, Newton não pode continuar")
            return x, i, aproximacoes, False

        x_novo = x - f(x) / df(x)
        aproximacoes.append(x_novo)

        if abs(x_novo - x) < tol or abs(f(x_novo)) < tol:
            return x_novo, i + 1, aproximacoes, True

        x = x_novo

    return x, max_iter, aproximacoes, False
```

```python
def secante(f, x0, x1, tol=TOL, max_iter=MAX_ITER):
    aproximacoes = [x0, x1]

    for i in range(max_iter):
        if abs(f(x1) - f(x0)) < 1e-12:
            print("Erro: denominador muito perto de zero, secante não pode continuar")
            return x1, i, aproximacoes, False

        x2 = x1 - f(x1) * (x1 - x0) / (f(x1) - f(x0))
        aproximacoes.append(x2)

        if abs(x2 - x1) < tol or abs(f(x2)) < tol:
            return x2, i + 1, aproximacoes, True

        x0, x1 = x1, x2

    return x1, max_iter, aproximacoes, False
```

### Teste rápido — a raiz de $x^2-2$ deve dar $\sqrt{2} \approx 1{,}41421356$

```python
f = lambda x: x**2 - 2
df = lambda x: 2*x

raiz, n_iter, aprox, convergiu = newton(f, df, 1.5)
print("raiz:", raiz, "| iterações:", n_iter, "| convergiu:", convergiu)
```

## Parte 2 — Aplicação nas funções propostas

Defino as quatro funções e suas derivadas (para Newton), e aplico os três métodos
em cada uma, usando os valores sugeridos no enunciado.

```python
def f1(x): return x**2 - 2
def df1(x): return 2*x

def f2(x): return x**3 - x - 2
def df2(x): return 3*x**2 - 1

def f3(x): return np.exp(-x) - x
def df3(x): return -np.exp(-x) - 1

def f4(x): return x**3 - 2*x + 2
def df4(x): return 3*x**2 - 2
```

```python
# f1(x) = x^2 - 2
raiz_b1, it_b1, aprox_b1, conv_b1 = bissecao(f1, 1, 2)
raiz_n1, it_n1, aprox_n1, conv_n1 = newton(f1, df1, 1.5)
raiz_s1, it_s1, aprox_s1, conv_s1 = secante(f1, 1, 2)

print("Bisseção:", raiz_b1, "em", it_b1, "iterações")
print("Newton:  ", raiz_n1, "em", it_n1, "iterações")
print("Secante: ", raiz_s1, "em", it_s1, "iterações")
```

```python
# f2(x) = x^3 - x - 2
raiz_b2, it_b2, aprox_b2, conv_b2 = bissecao(f2, 1, 2)
raiz_n2, it_n2, aprox_n2, conv_n2 = newton(f2, df2, 1.5)
raiz_s2, it_s2, aprox_s2, conv_s2 = secante(f2, 1, 2)

print("Bisseção:", raiz_b2, "em", it_b2, "iterações")
print("Newton:  ", raiz_n2, "em", it_n2, "iterações")
print("Secante: ", raiz_s2, "em", it_s2, "iterações")
```

```python
# f3(x) = e^-x - x
raiz_b3, it_b3, aprox_b3, conv_b3 = bissecao(f3, 0, 1)
raiz_n3, it_n3, aprox_n3, conv_n3 = newton(f3, df3, 0.5)
raiz_s3, it_s3, aprox_s3, conv_s3 = secante(f3, 0, 1)

print("Bisseção:", raiz_b3, "em", it_b3, "iterações")
print("Newton:  ", raiz_n3, "em", it_n3, "iterações")
print("Secante: ", raiz_s3, "em", it_s3, "iterações")
```

```python
# f4(x) = x^3 - 2x + 2
raiz_b4, it_b4, aprox_b4, conv_b4 = bissecao(f4, -2, -1)
raiz_n4, it_n4, aprox_n4, conv_n4 = newton(f4, df4, -2)
raiz_s4, it_s4, aprox_s4, conv_s4 = secante(f4, -2, -1)

print("Bisseção:", raiz_b4, "em", it_b4, "iterações")
print("Newton:  ", raiz_n4, "em", it_n4, "iterações")
print("Secante: ", raiz_s4, "em", it_s4, "iterações")
```

### Caso de falha (Newton com chute ruim)

Testei também Newton em $f_4$ começando em $x_0 = 0$. Aqui $f'(0) = -2$ (não é zero),
mas o método entra num ciclo: $x_0=0 \to x_1=1 \to x_2=0 \to \dots$ e nunca converge.
Isso mostra que Newton depende muito da escolha do chute inicial.

```python
raiz_falha, it_falha, aprox_falha, conv_falha = newton(f4, df4, 0)
print("convergiu?", conv_falha, "| depois de", it_falha, "iterações")
print("primeiras aproximações:", aprox_falha[:6])
```

## Parte 3 — Animações

Faço uma função de animação para cada método (são bem parecidas, mas prefiro deixar
separadas para ficar mais fácil de entender o que cada uma desenha).

```python
def animar_bissecao(f, a, b, aproximacoes, nome_arquivo):
    xs = np.linspace(a - 0.3, b + 0.3, 300)
    ys = f(xs)

    fig, ax = plt.subplots(figsize=(6, 4))

    # reconstrói os intervalos [a,b] de cada iteração
    intervalos = []
    ai, bi = a, b
    for m in aproximacoes:
        intervalos.append((ai, bi))
        if f(ai) * f(m) < 0:
            bi = m
        else:
            ai = m

    def desenha(i):
        ax.clear()
        ax.plot(xs, ys, "b")
        ax.axhline(0, color="black", linewidth=1)
        ai, bi = intervalos[i]
        ax.axvspan(ai, bi, color="orange", alpha=0.3)
        ax.plot(aproximacoes[i], 0, "ro", markersize=8)
        ax.set_title(f"Bisseção — iteração {i+1}\nx = {aproximacoes[i]:.6f} | f(x) = {f(aproximacoes[i]):.2e}")
        ax.grid(alpha=0.3)

    anim = FuncAnimation(fig, desenha, frames=len(aproximacoes))
    anim.save(nome_arquivo, writer=PillowWriter(fps=2))
    plt.close(fig)
```

```python
def animar_newton(f, df, x0, aproximacoes, nome_arquivo, xlim):
    xs = np.linspace(xlim[0], xlim[1], 300)
    ys = f(xs)

    fig, ax = plt.subplots(figsize=(6, 4))
    todos = [x0] + aproximacoes

    def desenha(i):
        ax.clear()
        ax.plot(xs, ys, "b")
        ax.axhline(0, color="black", linewidth=1)

        x_atual = todos[i]
        reta = f(x_atual) + df(x_atual) * (xs - x_atual)
        ax.plot(xs, reta, "r--")
        ax.plot(x_atual, f(x_atual), "ro")
        ax.plot(todos[i+1], 0, "go", markersize=8)

        ax.set_ylim(min(ys.min(), -1), max(ys.max(), 1))
        ax.set_title(f"Newton — iteração {i+1}\nx = {todos[i+1]:.6f} | f(x) = {f(todos[i+1]):.2e}")
        ax.grid(alpha=0.3)

    anim = FuncAnimation(fig, desenha, frames=len(aproximacoes)-1)
    anim.save(nome_arquivo, writer=PillowWriter(fps=1.5))
    plt.close(fig)
```

```python
def animar_secante(f, aproximacoes, nome_arquivo, xlim):
    xs = np.linspace(xlim[0], xlim[1], 300)
    ys = f(xs)

    fig, ax = plt.subplots(figsize=(6, 4))

    def desenha(i):
        ax.clear()
        ax.plot(xs, ys, "b")
        ax.axhline(0, color="black", linewidth=1)

        x0, x1 = aproximacoes[i], aproximacoes[i+1]
        reta = f(x1) + (f(x1) - f(x0)) / (x1 - x0) * (xs - x1)
        ax.plot(xs, reta, "g--")
        ax.plot([x0, x1], [f(x0), f(x1)], "go")
        ax.plot(aproximacoes[i+2], 0, "ro", markersize=8)

        ax.set_title(f"Secante — iteração {i+1}\nx = {aproximacoes[i+2]:.6f} | f(x) = {f(aproximacoes[i+2]):.2e}")
        ax.grid(alpha=0.3)

    anim = FuncAnimation(fig, desenha, frames=len(aproximacoes)-2)
    anim.save(nome_arquivo, writer=PillowWriter(fps=1.5))
    plt.close(fig)
```

## Conclusão

- A **bisseção** foi o método mais lento em todos os casos (por volta de 20 iterações),
  mas é o mais seguro: sempre converge desde que $f(a)$ e $f(b)$ tenham sinais opostos.
- O **Newton** foi o mais rápido (2 a 4 iterações), porque a convergência é quadrática.
  Em compensação precisa da derivada e é sensível ao chute inicial — no teste com
  $x_0=0$ em $f_4$ ele nem converge, fica preso num ciclo.
- A **secante** ficou no meio termo (4 a 6 iterações) e tem a vantagem de não precisar
  calcular a derivada.
